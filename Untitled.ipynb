```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import nltk
import matplotlib.pyplot as plt
import seaborn as sns

nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('punkt')

print("✅ All libraries loaded successfully!")
print("🚀 You are ready to build the Resume Screener!")
```

```python
import re
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Text cleaner
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

def clean_text(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    tokens = text.split()
    tokens = [lemmatizer.lemmatize(w) for w in tokens 
              if w not in stop_words and len(w) > 2]
    return " ".join(tokens)

# Skills dictionary
SKILLS = ["python", "machine learning", "deep learning", "sql",
          "nlp", "tensorflow", "pandas", "data analysis",
          "scikit-learn", "tableau", "power bi", "excel",
          "statistics", "numpy", "matplotlib", "seaborn"]

def extract_skills(text):
    return [skill for skill in SKILLS if skill in text.lower()]

print("✅ Preprocessing functions ready!")
```

```python
# Sample resumes (we'll use text directly — no PDF needed to start)

resumes = {
    "Alice_Resume": """
        Alice Johnson. Python developer with 2 years experience.
        Skills: Python, Machine Learning, Pandas, Scikit-learn, NLP, SQL, NumPy, Statistics.
        Experience: Data Analyst at ABC Corp. Built ML models for customer segmentation.
        Education: B.Tech Computer Science 2024.
    """,
    "Bob_Resume": """
        Bob Smith. Software engineer with Java and web development background.
        Skills: Java, JavaScript, HTML, CSS, MySQL, Excel.
        Experience: Web Developer at XYZ Ltd. Built e-commerce websites.
        Education: B.Tech Information Technology 2024.
    """,
    "Carol_Resume": """
        Carol Davis. Data science enthusiast with strong analytical skills.
        Skills: Python, Deep Learning, TensorFlow, NLP, Pandas, Data Analysis, Matplotlib, Seaborn, Statistics.
        Experience: ML Intern at DataCorp. Built NLP models for text classification.
        Education: B.Tech Computer Science 2025.
    """,
    "David_Resume": """
        David Lee. Business analyst with data visualization experience.
        Skills: Excel, Tableau, Power BI, SQL, Statistics, Data Analysis.
        Experience: Business Analyst at FinCorp. Created dashboards and reports.
        Education: B.Com Finance 2024.
    """,
    "Eva_Resume": """
        Eva Wilson. Machine learning engineer passionate about AI.
        Skills: Python, Machine Learning, Deep Learning, Scikit-learn, NumPy, Pandas, NLP, TensorFlow, SQL.
        Experience: AI Research Intern. Published paper on NLP text summarization.
        Education: B.Tech AI & ML 2025.
    """
}

# Job Description
job_description = """
    We are looking for a Data Scientist Intern with strong Python skills.
    Required: Machine Learning, NLP, Pandas, Scikit-learn, SQL, Data Analysis, Statistics, NumPy.
    Experience with data visualization and deep learning is a plus.
"""

# Clean all resumes and JD
cleaned_resumes = {name: clean_text(text) for name, text in resumes.items()}
cleaned_jd = clean_text(job_description)

print("✅ Resumes and Job Description loaded!")
print(f"📄 Total candidates: {len(resumes)}")
print("\n🔍 Sample cleaned resume (Alice):")
print(cleaned_resumes["Alice_Resume"])
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd

# Combine JD + all resumes into one list
all_docs = [cleaned_jd] + list(cleaned_resumes.values())
candidate_names = list(cleaned_resumes.keys())

# Build TF-IDF matrix
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(all_docs)

# Cosine similarity between JD (index 0) and each resume
scores = cosine_similarity(tfidf_matrix[0], tfidf_matrix[1:])[0]

# Build results dataframe
results = pd.DataFrame({
    "Candidate": candidate_names,
    "Match Score (%)": (scores * 100).round(2),
    "Skills Found": [extract_skills(text) for text in resumes.values()]
})

# Rank candidates
results = results.sort_values("Match Score (%)", ascending=False).reset_index(drop=True)
results["Rank"] = results.index + 1
results["Status"] = results["Match Score (%)"].apply(
    lambda x: "✅ Shortlisted" if x >= 50 else "❌ Not Selected"
)

print("=" * 55)
print("        CANDIDATE RANKING RESULTS")
print("=" * 55)
for _, row in results.iterrows():
    print(f"\nRank #{int(row['Rank'])} — {row['Candidate']}")
    print(f"  Match Score : {row['Match Score (%)']}%")
    print(f"  Status      : {row['Status']}")
    print(f"  Skills      : {', '.join(row['Skills Found'])}")
print("=" * 55)
```

```python
# Lower threshold to 30% since our resumes are short text samples
results["Status"] = results["Match Score (%)"].apply(
    lambda x: "✅ Shortlisted" if x >= 30 else "❌ Not Selected"
)

import matplotlib.pyplot as plt
import seaborn as sns

# ── Chart 1: Candidate Ranking Bar Chart ──
fig, ax = plt.subplots(figsize=(10, 6))

colors = ["#2ecc71" if s >= 30 else "#e74c3c" for s in results["Match Score (%)"]]
bars = ax.barh(results["Candidate"], results["Match Score (%)"], color=colors)

# Add score labels on bars
for bar, score in zip(bars, results["Match Score (%)"]):
    ax.text(bar.get_width() + 0.5, bar.get_y() + bar.get_height()/2,
            f'{score}%', va='center', fontsize=11, fontweight='bold')

ax.axvline(x=30, color='gray', linestyle='--', linewidth=1.5, label='Threshold (30%)')
ax.set_xlabel("Match Score (%)", fontsize=12)
ax.set_title("Resume Screening — Candidate Match Score", fontsize=14, fontweight='bold')
ax.legend()
ax.set_xlim(0, 55)
plt.tight_layout()
plt.savefig("outputs/ranking_chart.png", dpi=150)
plt.show()
print("✅ Chart 1 saved!")

# ── Chart 2: Skills Heatmap ──
skill_list = ["python", "machine learning", "deep learning", "sql",
              "nlp", "tensorflow", "pandas", "scikit-learn", "statistics"]

skill_matrix = pd.DataFrame(
    {skill: [1 if skill in extract_skills(text) else 0
             for text in resumes.values()]
     for skill in skill_list},
    index=candidate_names
)

fig, ax = plt.subplots(figsize=(12, 5))
sns.heatmap(skill_matrix, annot=True, cmap="YlGnBu",
            linewidths=0.5, cbar=False, ax=ax,
            fmt="d")
ax.set_title("Skills Presence Heatmap", fontsize=14, fontweight='bold')
ax.set_xlabel("Skills")
ax.set_ylabel("Candidates")
plt.tight_layout()
plt.savefig("outputs/skills_heatmap.png", dpi=150)
plt.show()
print("✅ Chart 2 saved!")

# ── Final Summary ──
print("\n" + "="*45)
print("         FINAL SHORTLIST")
print("="*45)
shortlisted = results[results["Status"] == "✅ Shortlisted"]
rejected = results[results["Status"] == "❌ Not Selected"]
print(f"✅ Shortlisted : {len(shortlisted)} candidates")
for _, r in shortlisted.iterrows():
    print(f"   #{int(r['Rank'])} {r['Candidate']} — {r['Match Score (%)']}%")
print(f"❌ Not Selected: {len(rejected)} candidates")
for _, r in rejected.iterrows():
    print(f"   #{int(r['Rank'])} {r['Candidate']} — {r['Match Score (%)']}%")
print("="*45)
```

```python
from wordcloud import WordCloud

# ── Chart 3: Word Cloud ──
all_resume_text = " ".join(resumes.values())

wc = WordCloud(
    width=800, height=400,
    background_color='white',
    colormap='viridis',
    max_words=50
).generate(all_resume_text)

plt.figure(figsize=(12, 6))
plt.imshow(wc, interpolation='bilinear')
plt.axis('off')
plt.title("Most Common Keywords Across All Resumes", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig("outputs/wordcloud.png", dpi=150)
plt.show()
print("✅ Word Cloud saved!")

# ── Save results to CSV ──
results_export = results[["Rank", "Candidate", "Match Score (%)", "Status"]].copy()
results_export["Skills Found"] = results["Skills Found"].apply(lambda x: ", ".join(x))
results_export.to_csv("outputs/screening_results.csv", index=False)
print("✅ Results saved to outputs/screening_results.csv")

print("\n🎯 PROJECT COMPLETE!")
print("="*45)
print("Files saved in outputs/ folder:")
print("  📊 ranking_chart.png")
print("  🔥 skills_heatmap.png")
print("  ☁️  wordcloud.png")
print("  📋 screening_results.csv")
print("="*45)
```
